# 💡 Project Proposal 1: Smart Study Companion – The Agentic Student Assistant


## 🧠 Problem Statement
Students often struggle to balance university lectures, exercises, and self-study with their personal life.  
Traditional calendar tools only schedule events – they don’t *understand* learning patterns, optimal repetition cycles, or cognitive load.  
As a result, students either undertrain (forgetting material) or overtrain (burnout).


## 🤖 Proposed Solution

We build an agentic AI assistant that automatically creates and adapts an *optimal learning plan* for the student.  
It combines calendar data, lecture and exam schedules, and scientific learning principles (e.g. spaced repetition, memory decay, and rest balance).  

The assistant:
- Reads the students calendar and identifies lectures, exercises, and exams.
- Calculates ideal repetition times for each lecture topic.
- Integrates hobbies, sports, and social activities.
- Creates a balanced schedule that maximizes learning efficiency and well-being.



## ⚙️ Core Agentic Design
**Agent Type:** Tool-based ReAct Agent (LLM with tools and continuous Thought–Action–Observation cycle)

**Key Tools:**
- 🗓️ `calendar_tool()` – reads and writes events (lectures, exams, hobbies)
- 📘 `learning_curve_tool()` – calculates optimal repetition intervals (Ebbinghaus forgetting curve)
- 🧩 `activity_balancer_tool()` – allocates breaks, hobbies, and social time
- 🕐 `time_optimizer_tool()` – simulates different week structures to maximize performance

**Thought–Action–Observation Loop Example:**
- **Thought:** “The student has an exam in 21 days and two lectures per week.”
- **Action:** Calls `learning_curve_tool()` → suggests reviews after 2, 5, and 12 days.
- **Observation:** Finds free slots → adds sessions to the calendar.
- **Thought:** “Student also has dance practice twice a week → avoid overlap.”
- **Final Answer:** Optimized weekly plan with personalized study and rest periods.


## 🧩 Architecture Overview
The agent is powered by:
- **ReAct Framework:** Combines reasoning (“how to schedule?”) with acting (calling calendar APIs).
- **Memory Module:** Tracks user feedback (e.g. “too much on Tuesday → reschedule”).
- **Graph-based Planning:** Nodes = tasks (lecture, repeat, rest), Edges = dependencies (lecture → repeat).

This allows adaptive scheduling and continuous optimization throughout the semester.


## 🌍 Expected Outcome
- A personalized, adaptive study plan updated weekly.
- Improved knowledge retention through spaced repetition.
- Better time management, reduced stress, and balanced lifestyle.
- Optional dashboard visualization for progress and workload balance.


## 🚀 Why It’s Strong for the Course
- Demonstrates **Agentic AI principles**: autonomy, proactivity, tool use, reasoning, feedback adaptation.
- Combines **cognitive science** and **AI planning** in a meaningful real-world application.
- Technically feasible for a working prototype using APIs (calendar, scheduling, LLM reasoning).


# 💡 Project Proposal 2: Individual Coach – The Personalized Fitness & Nutrition Agent


## 🧠 Problem Statement
People often struggle to maintain consistent fitness routines and balanced nutrition.  
Conventional apps provide generic meal or workout plans but lack true *personalization* — they don’t adapt to mental or physical progress, individual preferences, or emotional feedback.  

As a result, users lose motivation or follow unsustainable routines that don’t match their goals or lifestyle.


## 🤖 Proposed Solution
We design an **Agentic AI Individual Coach** that combines fitness, nutrition, and mental balance into one adaptive system.  
The agent acts as both a trainer and reflective partner, adjusting workouts, meals, and feedback according to the user’s current state.  

It performs:
1. **Initial Assessment:** age, fitness level, gender, and goals (e.g., strength, weight loss, endurance).  
2. **Personality Configuration:** user selects a preferred agent personality (strict, motivating, empathetic, calm).  
3. **Weekly Feedback Cycle:** the agent summarizes progress, identifies weak spots, and adapts next week’s plan.  
4. **Meal Planning:** the agent retrieves healthy meal options from recipe websites and creates a weekly menu.  
5. **Nutrient Tracking:** keeps track of calories, protein, fats, and carbohydrates.  
6. **Mental Fitness Check-ins:** asks reflective questions to support motivation and self-awareness.


## ⚙️ Core Agentic Design
**Agent Type:** Multi-tool ReAct Agent with persistent memory and dynamic personality

**Key Tools:**
- 🧮 `fitness_tool()` – calculates workout volume, intensity, and recovery based on progress
- 🥗 `nutrition_tool()` – connects to cooking websites and generates meal plans
- 📊 `tracking_tool()` – logs kcal, protein, and performance data
- 💬 `feedback_tool()` – conducts reflective weekly check-ins
- 🧠 `mood_tool()` – tracks emotional well-being via user feedback

**Thought–Action–Observation Loop Example:**
- **Thought:** “User reported fatigue and missed one workout.”
- **Action:** Adjust training plan → reduce volume, add recovery day.
- **Observation:** New performance metrics → improved adherence next week.
- **Thought:** “User prefers a more motivating tone.”
- **Final Answer:** “Let’s crush this week together! New plan uploaded 🚀.”


## 🧩 Architecture Overview
The system integrates:
- **ReAct Framework:** Combines reasoning and tool actions (e.g. data retrieval + adaptation).
- **Memory Component:** Stores user history, preferences, and feedback for continuity.
- **Graph-Based Reasoning:** Links nodes (Training, Nutrition, Mindset, Feedback) and dynamically adjusts based on dependencies.
- **Personality Module:** Alters agent responses (tone, strictness, motivational level) via system prompt changes.

This creates an adaptive, long-term coaching experience that evolves with the user.


## 🌍 Expected Outcome
- Fully personalized weekly plans for training and nutrition.
- Continuous progress tracking with meaningful feedback loops.
- Improved physical and mental health through reflection and adaptation.
- Dynamic agent personality fostering long-term engagement and motivation.


## 🚀 Why It’s Strong for the Course
- Demonstrates **all core Agentic AI principles**: autonomy, proactivity, reasoning, tool integration, memory, reflection, and feedback.
- Expands on the **Thought–Action–Observation** cycle by integrating emotional and cognitive adaptation.
- Technically feasible using APIs (e.g., fitness data, nutrition databases) and LLM-based reasoning.
- Excellent example of a **multi-domain intelligent agent** (sport, nutrition, psychology).


# 💡 Project Proposal 3: Data Guardian – An Autonomous Knowledge Graph Maintenance Agent


## 🧠 Problem Statement
Modern organizations rely on large internal knowledge bases — but they quickly become outdated or inconsistent.  
Employees duplicate data, references become stale, and factual errors propagate over time.  
Maintaining a consistent and up-to-date knowledge graph or database is costly and often done manually.


## 🤖 Proposed Solution
We develop an **autonomous “Data Guardian” Agent** that maintains and validates a company’s internal knowledge graph.  
The agent proactively detects outdated or inconsistent nodes and edges, fetches up-to-date information from trusted web or API sources, and performs self-checks before updating the knowledge base.

The agent acts like a librarian and detective combined:
- **Monitors** the knowledge graph for missing or inconsistent relationships.
- **Fetches** external data via APIs (e.g. Wikipedia, company APIs).
- **Validates** updates through reasoning (confidence scores, citation checks).
- **Documents** every change in a transparent, human-readable log.


## ⚙️ Core Agentic Design
**Agent Type:** Code + Tool Agent (ReAct + Graph-based workflow)

**Key Tools:**
- 🌐 `web_search_tool()` – retrieves factual data from reliable APIs.
- 🧩 `graph_validator_tool()` – checks node/edge consistency using graph rules.
- 📚 `update_tool()` – commits valid changes to the graph database.
- 🔍 `change_log_tool()` – generates human-readable summaries of all updates.
- 💾 `memory_module()` – keeps past changes for auditing and rollback.

**Thought–Action–Observation Loop Example:**
- **Thought:** “The edge (Product_A → Released_in → 2023) may be outdated.”
- **Action:** `web_search_tool("latest release date Product A")`
- **Observation:** “Found new release in 2025.”
- **Thought:** “Update confirmed from 3 sources, confidence 0.92.”
- **Action:** `update_tool(Product_A → Released_in → 2025)`
- **Final Answer:** “Knowledge graph successfully updated.”


## 🧩 Architecture Overview
- **Graph-Based Workflow:**  
  - Nodes = Entities (products, authors, concepts)  
  - Edges = Relations (released_in, developed_by, depends_on)
- **Directed Acyclic Graph (DAG)** ensures no update loops.
- **ReAct Framework:** Combines reasoning and verification steps.
- **Stop-and-Parse Approach:** Executes update actions in JSON format.
- **Feedback Mechanism:** The agent verifies its own updates and corrects mistakes in the next cycle.

This architecture allows the agent to evolve knowledge bases dynamically and autonomously.


## 🌍 Expected Outcome
- Real-time data consistency for any knowledge graph or semantic database.
- Reduced manual maintenance costs.
- Transparent audit trail of all modifications.
- Proof-of-concept for *self-healing data systems* powered by agentic AI.


## 🚀 Why It’s Strong for the Course
- Combines **core agentic mechanisms** (Thought–Action–Observation, ReAct, tool calls, graphs).
- Technically ambitious and nerdy: involves **graph databases**, **automated reasoning**, and **data validation**.
- Demonstrates deep understanding of **autonomy, proactivity, and feedback integration**.
- Perfect mix of theory (agent architecture) and practical application (data-driven AI).
